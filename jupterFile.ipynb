### Roll dice
class : Dice
method : roll and the roll should be a tuple

```python
import random
class Dice():
    def roll(self):
        first  = random.randint(1,6)
        second = random.randint(1,6)
        return first, second


dice1 = Dice()
dice1.roll()
```

### making new music model

```python
# 1. Import tools
import pandas as pd                    # For reading data
from sklearn.tree import DecisionTreeClassifier  # For ML model
# import joblib                 # For saving model
from sklearn import tree

music_data = pd.read_csv('music.csv')  # Read the CSV file


# 3. Prepare data
X = music_data.drop(columns=['genre']) # Features: Age, Gender
y = music_data['genre']                # Target: Genre


# 4. Create and train model
model = DecisionTreeClassifier()       # Create empty model
model.fit(X, y)            # Train it

tree.export_graphviz(model, out_file='music-recommender.dot', 
                     feature_names=['age', 'gender'], 
                     class_names=sorted(y.unique()), 
                     label='all', 
                     rounded=True, 
                     filled=True)  # Export decision tree to .dot file
# model = joblib.load('music-recommender.joblib')  # Load model from file

# # 5. Make predictions
# predictions = model.predict([[21, 1]])    # Predict for test data
# predictions2 = model.predict([[25,1]])    # Predict for test data
# predictions
```
