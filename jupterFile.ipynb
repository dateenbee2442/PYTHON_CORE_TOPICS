### Random access using random module


```python
import random

members = ['Shehi', 'Good boy', 'Jafar', 'Ingawa', 'Dateen']
leader = random.choice(members)
print(f'Among this list your leader is {leader}')
for i in members:
    print(i)
```

### Roll dice
class : Dice
method : roll and the roll should be a tuple

```python
import random
class Dice():
    def roll(self):
        first  = random.randint(1,6)
        second = random.randint(1,6)
        return first, second


dice1 = Dice()
dice1.roll()
```
