```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])

# DataFrame Copy
df_original = df.copy()
```

```python
df_altered = df_original.copy()
salary_year_avg = df_altered['salary_hour_avg'].max()
df_altered['salary_hour_avg'] = df_altered['salary_hour_avg'].fillna(salary_year_avg)
df_altered['salary_hour_avg'].sample(10)
```

using the .copy() the original dataset was'nt affected

```python
df_original['salary_hour_avg']
```

while the copied data was affected

```python
###using this kind of method saves you from altering the original data and not making changes that will change the result

df_altered['salary_hour_avg']
```

```python
salary_year_avg = df_altered['salary_year_avg'].min()
df_altered['salary_year_avg'] = df_altered['salary_year_avg'].fillna(salary_year_avg)
df_altered
```

pivoting_and_making_a_chart

```python
top_countries = df['job_country'].value_counts().head(5).index

job_country_salary = df.pivot_table(
    index='job_country',
    values='salary_year_avg',
    columns='job_title_short',
    aggfunc='median'
)

##Filters the pivot table to keep only the top 5 countries, .loc[top_countries] selects rows whose index labels match the countries in top_countries.
job_country_salary = job_country_salary.loc[top_countries]

job_titles = ['Data Analyst', 'Data Scientist', 'Data Engineer']
##Selects only those three columns from the pivot table. job_country_salary[job_titles] returns a DataFrame with only the columns in job_titles.
job_country_salary = job_country_salary[job_titles]
job_country_salary.plot(kind='bar')
plt.title('Median Salary by Country and Job Title')
plt.ylabel('Median Salary ($USD)')
plt.xlabel('')
plt.yticks(rotation=45, ha='right')
plt.xticks(rotation=45, ha='right')
plt.show()
                        
```

```python
top_countries = df['job_country'].value_counts().head(5).index

job_country_salary = df.pivot_table(
    index='job_country',
    values='salary_year_avg',
    columns='job_title_short',
    aggfunc='median'
).loc[top_countries]

job_titles = ['Data Analyst', 'Data Scientist', 'Data Engineer']
job_country_salary = job_country_salary[job_titles]

job_country_salary.plot(kind='bar', figsize=(10, 6))
plt.title('Median Salary by Country and Job Title')
plt.ylabel('Median Salary ($USD)')
plt.xlabel('')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show() 
```
