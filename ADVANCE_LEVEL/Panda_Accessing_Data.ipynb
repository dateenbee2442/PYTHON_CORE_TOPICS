```python
median_salary_year = df['salary_year_avg'].median()
```

```python
median_salary_hour = df['salary_hour_avg'].median()
```

### removing the NaN values from the column of salary_year and salary_hour avg


```python
df_filled = df
df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(median_salary_year)
df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(median_salary_hour)
```

To remove duplicates from the dataset

```python
df_unique = df_filled
df_unique =df_unique.drop_duplicates()
print('Length of original datasets ', len(df_filled))
print('Length of dropped duplicates datasets ', len(df_unique))
print('Length of datasets lost ', len(df_filled) - len(df_unique))
```

To remove duplicates by explicitly citing the column you want to

```python
df_unique = df_unique.drop_duplicates(subset=['job_title', 'company_name'])

print('Length of original datasets ', len(df_filled))
print('Length of dropped duplicates datasets ', len(df_unique))
print('Length of datasets lost ', len(df_filled) - len(df_unique))
```
